# Criptografia Assimétrica

A criptografia assimétrica é um método de criptografia que utiliza um par de chaves distintas. Diferente da criptografia simétrica, onde a mesma chave é usada para criptografar e descriptografar, na criptografia assimétrica uma chave é usada para criptografar e a outra para descriptografar.
- Esse sistema foi criado para resolver problemas de segurança da troca de chaves em canais não seguros.


#### Principais Características:

- **Chaves Diferentes:** Se uma cifra, a outra decifra.

- **Velocidade:** Lentidão em comparação à criptografia simétrica: Algoritmos assimétricos são consideravelmente mais lentos do que os simétricos. Isso ocorre devido à complexidade dos cálculos matemáticos envolvidos, como a fatoração de grandes números ou operações com curvas elípticas.

- **Tamanho das Chaves:** A criptografia assimétrica normalmente requer chaves maiores para oferecer o mesmo nível de segurança que a criptografia simétrica.
 - Por exemplo, uma chave RSA de 2048 bits oferece segurança semelhante a uma chave simétrica de 128 bits.
 - Chaves menores são possíveis com algoritmos como ECC (criptografia de curvas elípticas), que oferece o mesmo nível de segurança com chaves muito menores (256 bits de ECC são equivalentes a 3072 bits de RSA).

- **Segurança:** Garante confidencialidade, integridade, autenticidade e não-repúdio. Uma pessoa pode enviar informações criptografadas (com a chave pública), e apenas o destinatário (com a chave privada) pode descriptografar, mantendo a informação segura.

- **Principais de Algoritmos:**
 - RSA (Rivest-Shamir-Adleman),
 - ECC (Criptografia de Curvas Elípticas),
 - DSA (Digital Signature Algorithm).


#### Chave Pública e Chave Privada:

- **Chave Pública**: Pode ser distribuída livremente e usada por qualquer pessoa para criptografar uma mensagem ou verificar uma assinatura digital. Como o próprio nome sugere, ela não precisa ser protegida.

- **Chave Privada**: Esta deve ser mantida em segredo, pois é utilizada para descriptografar mensagens que foram criptografadas com a chave pública correspondente, ou para assinar digitalmente uma mensagem. Somente o detentor da chave privada tem a capacidade de realizar essas operações.

#### Geração do par de chaves
- As chaves RSA devem ser pelo menos de 2048 bits

````python
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import serialization
from cryptography.hazmat.primitives.asymmetric import rsa

private_key = rsa.generate_private_key(
                                        public_exponent=65537,
                                        key_size=3072,
                                        backend=default_backend(), )
public_key = private_key.public_key()
````

**Estrutura do Arquivo PEM**
- Codificação Base64: O formato PEM usa codificação Base64 para representar dados binários em um formato legível por humanos. Embora isso torne os arquivos mais portáveis e fáceis de manipular, ele aumenta o tamanho do arquivo em comparação ao formato binário puro.

- Cabeçalhos e Rodapés: O formato PEM inclui cabeçalhos e rodapés adicionais, como` -----BEGIN PRIVATE KEY-----` e `-----END PRIVATE KEY-----`, além de possíveis metadados sobre o tipo de chave ou a criptografia de proteção da chave.




##### Serializando (gravando) a chave em disco.
 - Formato: Privacy-Enhanced Mail (PEM)

````python
private_bytes = private_key.private_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PrivateFormat.PKCS8,
        encryption_algorithm=serialization.NoEncryption(), )

with open('private_key.pem', 'xb') as private_file:
  private_file.write(private_bytes)

public_bytes = public_key.public_bytes(
        encoding=serialization.Encoding.PEM,
        format=serialization.PublicFormat.SubjectPublicKeyInfo, )

with open('public_key.pem', 'xb') as public_file:
  public_file.write(public_bytes)
````

Visualizando o aquivo gerado:

````python
with open('public_key.pem', 'r') as arquivo:
    conteudo = arquivo.read()
    print(conteudo)
````

````python
with open('private_key.pem', 'r') as arquivo:
    conteudo = arquivo.read()
    print(conteudo)
````

Observações: O arquivo PEM da chave privada é maior que o da chave pública.
 -  Nos arquivos PEM de chaves privadas, muitas vezes a chave pública também está armazenada junto com a chave privada. Isso permite que o software que usa a chave possa derivar a chave pública a partir da privada, mas o contrário não é possível. Portanto, o arquivo da chave privada inclui tanto a chave privada quanto a pública, o que já aumenta seu tamanho.

- Parâmetros Adicionais: Para algoritmos como o RSA, o arquivo PEM da chave privada contém não apenas o número primo utilizado, mas também vários parâmetros matemáticos que auxiliam no processo de criptografia e decriptação. Esses parâmetros incluem:

 - Modulus (n)
 - Exponentes públicos e privados (e e d)
 - Fatores primos (p e q)
 - Outros valores derivados como dP, dQ e qInv (relacionados à otimização de operações).

- A chave pública é composta principalmente por dois componentes: o modulus (n) e o exponente público (e). Isso resulta em um arquivo PEM relativamente pequeno.


##### Deserializando

````python
with open('private_key.pem', 'rb') as private_file:
  loaded_private_key = serialization.load_pem_private_key(
                            private_file.read(),
                            password=None,
                            backend=default_backend()
                          )

with open('public_key.pem', 'rb') as public_file:
  loaded_public_key = serialization.load_pem_public_key(
                          public_file.read(),
                          backend=default_backend()
                         )
````

#### Criptografando dos dados com RSA
Assim como cifradores de bloco simétricos, o RSA criptografa dados com um esquema de preenchimento (padding).

-  OAEP (Optimal Asymmetric Encryption Padding):
  
  O esquema recomendado para o RSA é o OAEP (Optimal Asymmetric Encryption Padding), que é uma técnica específica para melhorar a segurança durante os processos de criptografia e descriptografia com RSA. Essa técnica torna os dados resistentes a ataques, introduzindo aleatoriedade no processo.
Garante que mensagens iguais resultem em textos cifrados diferentes Ele usa funções de hash para gerar máscaras que embaralham a mensagem original e o número aleatório, o que impede ataques que dependem da repetição de padrões.

````python
from cryptography.hazmat.primitives import hashes
from cryptography.hazmat.primitives.asymmetric import padding

padding_config = padding.OAEP(
                    mgf=padding.MGF1(algorithm=hashes.SHA256()),
                    algorithm=hashes.SHA256(),
                    label=None, )

plaintext = b'Bob, este eh um segredo. Nao conte para ninguem. Ass: Alice'

ciphertext = loaded_public_key.encrypt(
                    plaintext=plaintext,
                    padding=padding_config, )
````

````python
print("Texto cifrado (hex):", ciphertext.hex())
print("Tamanho:", len(ciphertext), "bytes" )
print("Tamanho:", len(ciphertext)*8, "bits (!)" )
````

#### Decriptografando dos dados com RSA

````python
decrypted_by_private_key = loaded_private_key.decrypt(
                            ciphertext=ciphertext,
                            padding=padding_config)

print("Texto descriptografado (hex):", decrypted_by_private_key.hex())
print("Tamanho:", len(decrypted_by_private_key), "bytes" )
print('Text:', decrypted_by_private_key)
````

#### Cifrando mesmo texto mais vezes. Mesmo resultado?

````python
ciphertext1 = loaded_public_key.encrypt(
                    plaintext=plaintext,
                    padding=padding_config, )

ciphertext2 = loaded_public_key.encrypt(
                    plaintext=plaintext,
                    padding=padding_config, )

ciphertext3 = loaded_public_key.encrypt(
                    plaintext=plaintext,
                    padding=padding_config, )

print("Texto cifrado 1 (hex):", ciphertext1.hex())
print("Texto cifrado 2 (hex):", ciphertext2.hex())
print("Texto cifrado 3 (hex):", ciphertext3.hex())
````

````python
decrypted1 = loaded_private_key.decrypt(
                            ciphertext=ciphertext1,
                            padding=padding_config)
print('Text:', decrypted1)

decrypted2 = loaded_private_key.decrypt(
                            ciphertext=ciphertext2,
                            padding=padding_config)
print('Text:', decrypted2)

decrypted3 = loaded_private_key.decrypt(
                            ciphertext=ciphertext3,
                            padding=padding_config)
print('Text:', decrypted3)
````

####Como funciona o OAEP
O OAEP é baseado em funções de hash e aplica uma combinação de preenchimento aleatório e mistura de dados antes de criptografar a mensagem com RSA. O processo ocorre em várias etapas:

1. Entrada
- M: Mensagem a ser criptografada.
- k: Tamanho do bloco de saída do algoritmo RSA.
- G e H: Funções de hash (por exemplo, SHA-256) utilizadas para misturar a mensagem e o preenchimento.
- r: Um número aleatório (também chamado de "semente") gerado para cada operação de criptografia.
2. Preenchimento
- A mensagem M é preenchida até um tamanho fixo com bits extras, tornando-a adequada para ser criptografada por RSA. O objetivo é evitar que mensagens pequenas levem à vulnerabilidade ao usar criptografia RSA direta.
3. Geração de Máscara
- O número aleatório r é usado como entrada para a função de hash G, gerando uma máscara para a mensagem preenchida.
- A mensagem preenchida é então misturada com essa máscara por meio de uma operação XOR (ou exclusivo bit a bit).
4. Hash da Mensagem Preenchida
- O resultado da mistura é usado como entrada para outra função de hash H, que gera uma máscara para o número aleatório r.
- O número aleatório r é então misturado com essa nova máscara, também usando XOR.
5. Criptografia com RSA
- O resultado das operações de máscara é combinado e o bloco resultante é criptografado com RSA.
- Isso gera o código cifrado, que pode então ser transmitido ou armazenado.
6. Descriptografia (Processo Reverso)
- Na descriptografia, o receptor inverte o processo, aplicando a chave privada RSA para reverter a criptografia. Em seguida, ele usa as mesmas funções de hash G e H para recuperar o número aleatório r e, em última instância, a mensagem original M.

#### Vantagens do OAEP:
1. Aleatoriedade: O uso do número aleatório r garante que mensagens iguais resultem em textos cifrados diferentes, o que impede ataques que dependem da repetição de padrões.

2. Proteção contra Ataques: OAEP protege contra ataques de texto cifrado escolhido, como ataques de "malleability", onde o atacante pode modificar um texto cifrado e observar como a alteração afeta o texto plano após a descriptografia.

3. Segurança: As funções de hash G e H são usadas para garantir que as operações de máscara sejam seguras, prevenindo que um atacante consiga prever ou manipular os dados durante a criptografia.

**Curioso?**
##### Por que expoente 6655375?
```
private_key = rsa.generate_private_key(
                                        public_exponent=65537,
                                        key_size=3072,
                                        backend=default_backend(), )
```

O valor 65537 é amplamente utilizado como o expoente público em criptografia RSA por uma combinação de fatores técnicos e históricos. Ele oferece uma boa combinação de eficiência, segurança e compatibilidade. Por que?

- É um número primo: A escolha de um número primo para o expoente público simplifica os cálculos matemáticos envolvidos no algoritmo RSA, especificamente na geração e verificação das chaves.

- Tem poucas bits 1 em sua representação binária: O número 65537 tem apenas duas bits iguais a 1 em sua representação binária (10000000000000001), o que reduz significativamente o número de operações de multiplicação necessárias para executar a criptografia e a verificação de assinatura. Isso faz com que as operações com chaves públicas sejam mais rápidas e eficientes.

- O expoente 65537 é grande o suficiente para ser seguro, mas ainda pequeno o suficiente para não tornar os cálculos de criptografia e assinatura excessivamente lentos.

# Exercício:

**Construção de um Envelope Digital:**

Um envelope digital é um método de criptografia que combina técnicas de criptografia simétrica e assimétrica para proteger dados de forma segura e eficiente.
- Construa as funções:
   - Cifrar e construir o envelope digital
   - Decifrar envelope digital

- Utilize os algoritmos AES e RSA.
- Teste com mensagens com tamanhos diferentes e com chaves diferentes.



````python
import os
from cryptography.hazmat.backends import default_backend
from cryptography.hazmat.primitives import hashes, serialization
from cryptography.hazmat.primitives.asymmetric import rsa, padding
from cryptography.hazmat.primitives.ciphers.aead import AESGCM

# ==========================================
# 1. FUNÇÃO: CIFRAR E CONSTRUIR ENVELOPE
# ==========================================
def criar_envelope_digital(mensagem: bytes, chave_publica_rsa) -> dict:
    """
    Criptografa a mensagem usando AES-256-GCM e criptografa a chave simétrica com a chave pública RSA.
    """
    # 1. Gerar uma chave simétrica aleatória para o AES (256 bits / 32 bytes)
    chave_simetrica_aes = AESGCM.generate_key(bit_length=256)

    # 2. Criptografar a mensagem com AES-GCM
    aesgcm = AESGCM(chave_simetrica_aes)
    nonce = os.urandom(12)  # Vetor de inicialização / Nonce de 96 bits recomendado para GCM
    mensagem_cifrada = aesgcm.encrypt(nonce, mensagem, associated_data=None)

    # 3. Criptografar a chave simétrica AES usando a chave pública RSA (OAEP)
    padding_rsa = padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
    chave_aes_cifrada = chave_publica_rsa.encrypt(chave_simetrica_aes, padding_rsa)

    # 4. Retornar a estrutura do Envelope Digital
    envelope = {
        "chave_aes_cifrada": chave_aes_cifrada,
        "nonce": nonce,
        "mensagem_cifrada": mensagem_cifrada
    }
    return envelope

# ==========================================
# 2. FUNÇÃO: DECIFRAR ENVELOPE DIGITAL
# ==========================================
def decifrar_envelope_digital(envelope: dict, chave_privada_rsa) -> bytes:
    """
    Decriptografa a chave simétrica usando a chave privada RSA e recupera a mensagem original.
    """
    # 1. Decriptografar a chave simétrica AES usando a chave privada RSA
    padding_rsa = padding.OAEP(
        mgf=padding.MGF1(algorithm=hashes.SHA256()),
        algorithm=hashes.SHA256(),
        label=None
    )
    chave_simetrica_aes = chave_privada_rsa.decrypt(
        envelope["chave_aes_cifrada"],
        padding_rsa
    )

    # 2. Decriptografar a mensagem usando a chave simétrica AES recuperada
    aesgcm = AESGCM(chave_simetrica_aes)
    mensagem_original = aesgcm.decrypt(
        envelope["nonce"],
        envelope["mensagem_cifrada"],
        associated_data=None
    )

    return mensagem_original

# ==========================================
# 3. TESTES COM DIFERENTES TAMANHOS E CHAVES
# ==========================================
def gerador_chaves_rsa(tamanho_bits=2048):
    privada = rsa.generate_private_key(
        public_exponent=65537,
        key_size=tamanho_bits,
        backend=default_backend()
    )
    publica = privada.public_key()
    return privada, publica

print("--- INICIANDO TESTES DO ENVELOPE DIGITAL ---\n")

# Teste 1: Mensagem Pequena com Chave RSA de 2048 bits
print(">>> TESTE 1: Mensagem curta | Chave RSA 2048 bits")
priv_1, pub_1 = gerador_chaves_rsa(2048)
msg_1 = b"Mensagem curta de teste."

envelope_1 = criar_envelope_digital(msg_1, pub_1)
msg_decifrada_1 = decifrar_envelope_digital(envelope_1, priv_1)

print(f"Mensagem Original  : {msg_1.decode('utf-8')}")
print(f"Chave AES Cifrada (hex): {envelope_1['chave_aes_cifrada'].hex()[:40]}...")
print(f"Mensagem Cifrada  (hex): {envelope_1['mensagem_cifrada'].hex()}")
print(f"Mensagem Decifrada: {msg_decifrada_1.decode('utf-8')}")
print(f"Sucesso: {msg_1 == msg_decifrada_1}\n")

# Teste 2: Mensagem Longa com Chave RSA de 3072 bits
print(">>> TESTE 2: Mensagem longa | Chave RSA 3072 bits")
priv_2, pub_2 = gerador_chaves_rsa(3072)
msg_2 = (
    b"Este eh um texto muito mais longo para demonstrar que o envelope digital "
    b"supera a limitacao de tamanho do RSA! " * 50
)

envelope_2 = criar_envelope_digital(msg_2, pub_2)
msg_decifrada_2 = decifrar_envelope_digital(envelope_2, priv_2)

print(f"Tamanho da mensagem original: {len(msg_2)} bytes")
print(f"Tamanho da mensagem cifrada : {len(envelope_2['mensagem_cifrada'])} bytes")
print(f"Sucesso ao decifrar: {msg_2 == msg_decifrada_2}\n")

# Teste 3: Tentativa de decifrar com chave incorreta (Tratamento de erro)
print(">>> TESTE 3: Segurança (Tentativa de abrir com chave errada)")
try:
    decifrar_envelope_digital(envelope_1, priv_2)
except Exception as e:
    print(f"Falha esperada ao usar chave privada incorreta: {type(e).__name__}")
````
